```python

import pandas as pd

# Load the PipeGuard dataset
df = pd.read_csv("../PipeGuard_Main_Dataset.csv")

```

```python
# Display the first 5 rows
df.head()
```

```python

# Check the number of rows and columns
print("Dataset Shape:", df.shape)

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python

# Display all column names
print("Column Names:")
print(df.columns.tolist())

# Display dataset information
print("\nDataset Information:")
df.info()
```

```python

# Check missing values in each column
missing_values = df.isnull().sum()

# Display missing values
print("Missing Values:")
print(missing_values)
```

```python

# Count duplicate records
duplicates = df.duplicated().sum()

print("Total Duplicate Rows:", duplicates)
```

```python

# Calculate missing values and percentages
missing_count = df.isnull().sum()
missing_percent = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
"Missing Count": missing_count,
"Missing Percentage": missing_percent.round(2)
})

# Show columns with missing values
missing_summary[missing_summary["Missing Count"] > 0].sort_values(
by="Missing Percentage", ascending=False
)
```

```python

# Check exact duplicate rows
duplicate_count = df.duplicated().sum()

print("Total Rows:", len(df))
print("Exact Duplicate Rows:", duplicate_count)
print("Duplicate Percentage:",
round(duplicate_count / len(df) * 100, 2), "%")
```

```python


# Check unique values in categorical columns
categorical_columns = df.select_dtypes(include=["object"]).columns

for col in categorical_columns:
    print(f"\nColumn: {col}")
    print("Unique Values:", df[col].nunique())
    
    if col not in ["assessment_date", "installation_date"]:
        print(df[col].value_counts(dropna=False).head(20))

```

```python

# Check invalid negative or zero values
columns_to_check = [
    "pipe_length_m",
    "burial_depth_m",
    "diameter_mm",
    "pipe_age_years"
]

for col in columns_to_check:
    invalid_count = (df[col] <= 0).sum()
    print(f"{col}: {invalid_count} zero or negative values")
```
